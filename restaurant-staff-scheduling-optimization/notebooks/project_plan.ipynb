# Restaurant Staff Scheduling Optimization — Project Context for Codex

## Project Overview

This project is a professional end-to-end AI and operations optimization system for restaurant workforce planning.

The goal is to build a realistic restaurant staff scheduling system that combines:

* real restaurant operational data
* demand forecasting
* workforce scheduling
* operational constraints
* optimization logic
* business analytics

The project should simulate how a real restaurant could use AI and data-driven decision-making to optimize staffing operations.

This is NOT just a machine learning notebook project.
It should feel like a real operational AI system.

---

# Main Goals

The project should:

1. Analyze restaurant operational data
2. Understand customer demand patterns
3. Predict staffing needs
4. Generate optimized shift schedules
5. Balance:

   * labor cost
   * staffing coverage
   * employee availability
   * fairness
   * operational efficiency

---

# Project Theme

The project is inspired by real restaurant operations.

The system should answer questions like:

* How many chefs and servers are needed per shift?
* Which days require more staff?
* How should shifts be assigned?
* How can labor cost be reduced without hurting operations?
* How can schedules remain fair between workers?

# SYNTHETIC DATASETS

The synthetic datasets were intentionally created with dirty data to simulate real-world operational problems.

---

# staff.csv

Purpose:
Restaurant employee master table.

Contains:

* staff IDs
* names
* roles
* experience levels
* hourly costs
* contracts
* preferred shifts

Important:
This dataset intentionally includes:

* missing values
* invalid dates
* inconsistent categories
* duplicate IDs
* wrong data types
* unrealistic values

The cleaning process is part of the project.

---

# availability.csv

Purpose:
Daily employee availability data.

Contains:

* work date
* staff ID
* availability
* preferred shift
* availability reason

This dataset is critical for scheduling optimization.

It intentionally includes:

* inconsistent yes/no values
* invalid dates
* missing values
* invalid staff IDs
* duplicates
* inconsistent shift names

---

# daily_staff_requirements.csv

Purpose:
Bridge dataset between demand forecasting and optimization.

Contains:

* date
* shift
* expected orders
* expected sales
* required chefs
* required servers
* estimated labor budget

This dataset simulates operational staffing demand.

It intentionally includes:

* invalid staffing values
* dirty categorical values
* invalid dates
* missing values
* unrealistic sales values
* duplicates
